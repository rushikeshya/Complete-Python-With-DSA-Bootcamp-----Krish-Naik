# Flowchart and Problem Solving

#### Introduction to Flowcharts
1. A flowchart is a graphical representation of a process, system or algorithm.
2. It uses symbols and arrows to show the squence of step needed to complete a task.
3. Flow charts are useful in visualising how to solve a problem and make the process easier to follow.

#### What is Pseudocode?
1. A `Pseudocode` is a simple way of describing an algorithm in plain language of a mix of code like and natural language instructions.
2. It doesn't follow any specific programming language syntax, making it easier to understand, even for those without programming knowledge
3. Pseudo code is often used as a step before actual code.



```python
'''
Begin
    Input First Number
    Input Second Number

    SUM = FirstNumber + SecondNumber

    print SUM

END
'''
```

#### Framework to Solve Problem
1. Analyse our problem - Input, Output, Constraint
2. Break down probleam into smaller subparts.
3. Remember/enlist the concept.
4. Take 2-3 example  - Clear out confusion
5. Write a pseudocode on a paper.
6. Dry run it
7. Write down your solution.
8. Look Out for edge cases/boundory condition.
