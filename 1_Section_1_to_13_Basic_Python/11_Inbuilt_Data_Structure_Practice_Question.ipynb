#### Que 21. Sum of List Elements

Problem Description

Sum of List Elements

Write a Python function that calculates the sum of all elements in a given list of integers.

Parameters:

numbers (List of integers): The input list containing integers.

Returns:

An integer representing the sum of all elements in the input list.

Example:

Input: numbers = [1, 2, 3, 4, 5]
Output: 15

Input: numbers = [10, -5, 7, 8, -2]
Output: 18

```python
# Sum of list using sum() method
def sum_list(lst):
    return sum(lst)

sum_list([1,2,3,4,5])
```

```python
# Sum of list using for loop
def sum_list1(lst):
    sum = 0
    for i in lst:
        sum = sum + i  # sum += i
    return sum

sum_list1([1,2,3,4,5])
```

```python
# Sum of list using while loop
def sum_while(lst):
    total = 0
    i = 0
    while i < len(lst):
        total += lst[i]
        i += 1
    return total
print(sum_while([1,2,3,4,5]))
```

```python
# Sum of list using Recursion
def sum_recursion_list(lst):
    if not lst:
        return 0
    return lst[0] + sum_recursion_list(lst[1:])
print(sum_recursion_list([1,2,3,4,5]))
```

```python
## Using list comprehension
def sum_list_compr(lst):
    return sum([x for x in lst])
sum_list_compr([1,2,3,4,5])
```

#### 22. Largest Element in a List
Find the Largest Element in a List

Write a Python function that finds and returns the largest element in a given list of integers.

Parameters:

numbers (List of integers): The input list containing integers.

Returns:

An integer representing the largest element in the input list.

Example:

Input: numbers = [3, 8, 2, 10, 5]
Output: 10

Input: numbers = [-5, -10, -2, -1, -7]
Output: -1

```python
# Using simple max() inbuilt methods/function
def large_ele(numbers):
    return max(numbers)

num_list = [3,8,2,10,5]
large_ele(num_list)
```

```python
## Using loop (most common)
def larg_num_loop(lst):
    largest = lst[0]
    for num in lst:
        if num > largest:
            largest = num
    return largest
num_list = [3,8,2,10,5]
large_ele(num_list)
```

```python
## using sorting
def large_num_sorted(lst):
    return sorted(lst)[-1]
large_num_sorted(num_list)
```

```python
## using while loop
def large_ele_while(lst):
    i = 0
    largest = lst[0]
    while i < len(lst):
        if lst[i] > largest:
            largest = lst[i]
        i += 1
    return largest
large_ele_while(num_list)
```

Remove Duplicate in a List
Remove Duplicates from a List

You are given a list of integers. Write a Python program that removes any duplicate elements from the list and returns a new list with only unique elements. The order of elements in the list should be maintained.

Parameters:

lst (List of integers): The list of integers from which duplicates should be removed.

Returns:

A list of integers where all duplicates have been removed, preserving the original order.

Example:

Input: lst = [1, 2, 2, 3, 4, 4, 5]
Output: [1, 2, 3, 4, 5]

Input: lst = [4, 5, 5, 4, 6, 7]
Output: [4, 5, 6, 7]


Note :- Please don't use set or inbuilt functions. Try to use brute force logic.

```python
def remove_dupli(lst):
    unique = []
    for num in lst:
        already_present = False
        for u in unique:
            if u == num:
                already_present = True
                break
        # If not found, Append it
        if not already_present:
            unique.append(num)
    return unique

lst = [1, 2, 2, 3, 4, 4, 5]
remove_dupli(lst)
```
