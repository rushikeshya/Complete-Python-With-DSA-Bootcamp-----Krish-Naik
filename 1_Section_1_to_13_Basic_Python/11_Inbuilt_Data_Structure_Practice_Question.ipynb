#### Que 21. Sum of List Elements

Problem Description

Sum of List Elements

Write a Python function that calculates the sum of all elements in a given list of integers.

Parameters:

numbers (List of integers): The input list containing integers.

Returns:

An integer representing the sum of all elements in the input list.

Example:

Input: numbers = [1, 2, 3, 4, 5]
Output: 15

Input: numbers = [10, -5, 7, 8, -2]
Output: 18

```python
# Sum of list using sum() method
def sum_list(lst):
    return sum(lst)

sum_list([1,2,3,4,5])
```

```python
# Sum of list using for loop
def sum_list1(lst):
    sum = 0
    for i in lst:
        sum = sum + i  # sum += i
    return sum

sum_list1([1,2,3,4,5])
```

```python
# Sum of list using while loop
def sum_while(lst):
    total = 0
    i = 0
    while i < len(lst):
        total += lst[i]
        i += 1
    return total
print(sum_while([1,2,3,4,5]))
```

```python
# Sum of list using Recursion
def sum_recursion_list(lst):
    if not lst:
        return 0
    return lst[0] + sum_recursion_list(lst[1:])
print(sum_recursion_list([1,2,3,4,5]))
```

```python
## Using list comprehension
def sum_list_compr(lst):
    return sum([x for x in lst])
sum_list_compr([1,2,3,4,5])
```
