# Variable Length Arguments
## Positional And Keyword Arguments

```python
## Positinoal Arguments

def print_number(*args):
    for number in args:
        print(number)
```

```python
print_number(1,2,3,4,5,6,7,8,9,"rush")
```

```python
## keyword Arguments - All the parameters in the form of key-values pairs 

def print_details(**kwargs):
    for key,value in kwargs.items():
        print(f"{key}:{value}")

```

```python
print_details(name="Rushi", age=25, country="India")
```

```python
## Mix
def print_details(*args, **kwargs):
    for val in args:
        print(f"Positional arguments: {val}")
    
    for key, value in kwargs.items():
        print(f"{key}:{value}")
```

```python
print_details(1,2,3,4,5,6,2.5,6.4)
```

```python
print_details(name="Rushi", Age=25, Country="India")
```

# More Coding Example with Function

## Example 1: Temperature Conversion

```python
def temp_in_fahrenheit(temp):
    print(f"{temp * 9/5 + 32} F")
```

```python
temp_in_fahrenheit(30)
```

```python
def temp_in_celsius(temp):
    print(f"{(temp-32)*5/9:.2f} C")
```

```python
temp_in_celsius(80)
```

```python
def convert_temperature(temp, unit):
    """This Function convert temperature between Celsius and Fahrenheit"""
    if unit == 'C':
        return (temp * 9/5) + 32  ## Celsius to Fahrenheit
    elif unit == 'F':
        return (temp - 32) * 5/9  ## Fahrenheit to Celsius
    else:
        return None
result = convert_temperature(30,'C')
print(result)
```

```python
convert_temperature(102,"F")
```

#### Example 3: Password Strength Checker

```python
def pass_checker(password):
    if len(password) < 6:
        print("Week Password")
    elif len(password) > 6 and len(password) < 10:
        print("Medium Password")
    else:
        print("Strong")
```

```python
pass_checker("6561867")
```

```python
def is_strong_password(password):
    """This Function checks if the password is strong or not"""
    if len(password)<8:
        return False
    if not any(char.isdigit() for char in password):
        return False
    if not any(char.isupper() for char in password):
        return False
    if not any(char.islower() for char in password):
        return False
    return True
```

```python
is_strong_password("Rush@123")
```

### Example 3: Calculate the Total Cost of Items In a Shopping

```python
def calculate_total_cost(cart):
    total_cost = 0
    for item in cart:
        total_cost += item['price']*item['quantity']
    return total_cost

## Example of cart data
cart = [
    {'name':'Apple', 'price':0.5, 'quantity':4},
    {'name':'Banana', 'price':0.3, 'quantity':6},
    {'name':'Orange', 'price':0.7, 'quantity':3}
]

# Calling the function
total_cost = calculate_total_cost(cart)
print(total_cost)
```

#### Example 4: Check If a String Is Palindrome

```python
def is_palindrome(s):
    s = s.lower().replace(" ","")
    return s == s[::-1]

print(is_palindrome("A man a plan a canal Panama"))
print(is_palindrome("Adda"))
print(is_palindrome("Hello"))
```

#### Example 5: Calculate the function of a number using recursion

```python
def factorial(n):
    if n==0:
        return 1
    else:
        return n * factorial(n-1)
print(factorial(5))
print(factorial(4))
print(factorial(3))
```

#### Example 6: A Function to read a file and count the frequency of each word

```python
def count_word_frequency(file_path):
    word_count = {}
    with open(file_path, 'r') as file:
        for line in file:
            words = line.split()
            for word in words:
                word=word.lower().strip('.,!?:"\'')
                word_count[word]=word_count.get(word,0)+1
    return word_count
```

```python
count_word_frequency("sample.txt")
```

#### Valid Email Address

```python
import re

# Email validation function
def is_valid_email(email):
    """This function check if the email is valid."""
    pattern = r'^[a-zA-Z0-9_.+-]+@[a-zA-Z0-9-.]+$'
    return re.match(pattern, email) is not None

# Calling the function
print(is_valid_email("test@example.com")) #o/p: True
print(is_valid_email("invalid-email")) #output: False
```
