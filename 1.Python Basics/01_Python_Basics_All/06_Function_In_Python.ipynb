# Variable Length Arguments
## Positional And Keyword Arguments

```python
## Positinoal Arguments

def print_number(*args):
    for number in args:
        print(number)
```

```python
print_number(1,2,3,4,5,6,7,8,9,"rush")
```

```python
## keyword Arguments - All the parameters in the form of key-values pairs 

def print_details(**kwargs):
    for key,value in kwargs.items():
        print(f"{key}:{value}")

```

```python
print_details(name="Rushi", age=25, country="India")
```

```python
## Mix
def print_details(*args, **kwargs):
    for val in args:
        print(f"Positional arguments: {val}")
    
    for key, value in kwargs.items():
        print(f"{key}:{value}")
```

```python
print_details(1,2,3,4,5,6,2.5,6.4)
```

```python
print_details(name="Rushi", Age=25, Country="India")
```

# More Coding Example with Function

## Example 1: Temperature Conversion

```python
def temp_in_fahrenheit(temp):
    print(f"{temp * 9/5 + 32} F")
```

```python
temp_in_fahrenheit(30)
```

```python
def temp_in_celsius(temp):
    print(f"{(temp-32)*5/9:.2f} C")
```

```python
temp_in_celsius(80)
```

```python
def convert_temperature(temp, unit):
    """This Function convert temperature between Celsius and Fahrenheit"""
    if unit == 'C':
        return (temp * 9/5) + 32  ## Celsius to Fahrenheit
    elif unit == 'F':
        return (temp - 32) * 5/9  ## Fahrenheit to Celsius
    else:
        return None
result = convert_temperature(30,'C')
print(result)
```

```python
convert_temperature(102,"F")
```

#### Example 3: Password Strength Checker

```python
def pass_checker(password):
    if len(password) < 6:
        print("Week Password")
    elif len(password) > 6 and len(password) < 10:
        print("Medium Password")
    else:
        print("Strong")
```

```python
pass_checker("6561867")
```

```python
def is_strong_password(password):
    """This Function checks if the password is strong or not"""
    if len(password)<8:
        return False
    if not any(char.isdigit() for char in password):
        return False
    if not any(char.isupper() for char in password):
        return False
    if not any(char.islower() for char in password):
        return False
    return True
```

```python
is_strong_password("Rush@123")
```
